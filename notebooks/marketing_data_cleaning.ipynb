```python
url = "https://raw.githubusercontent.com/Lulukia0704/digital-marketing-campaign-analysis/main/data/Marketing_campaign.csv"

df = pd.read_csv(url)
```

```python
df['time'] = pd.to_datetime(df['time'], errors='coerce')
```

```python
cols_numeric = [
    'impressions',
    'clicks',
    'media_cost_usd',
    'campaign_budget_usd',
    'approved_budget'
]

for col in cols_numeric:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
cols_id = [
    'campaign_item_id',
    'ext_service_id',
    'creative_id',
    'advertiser_id',
    'network_id',
    'channel_id'
]

for col in cols_id:
    df[col] = df[col].astype(str)
```

```python
cols_drop = [
    'position_in_content',
    'unique_reach',
    'total_reach'
]

df = df.drop(columns=cols_drop)
```

```python
cols_drop = [
    'creative_id',
    'creative_width',
    'creative_height',
    'template_id',
    'search_tags',
    'search_tag_cat',
    'keywords',
    'approved_budget',
    'advertiser_currency',
    'currency_code',
    'stats_currency',
    'exchange_rate',
    'ext_service_id',
    'advertiser_id',
    'network_id',
    'channel_id',
    'max_bid_cpm',
    'network_margin',
    'landing_page',
    'timezone',
    'cmi_currency_code'
]
```

```python
# Konfigurasi GitHub
GITHUB_TOKEN = userdata.get('GITHUB_TOKEN')
OWNER = 'Lulukia0704'
REPO = 'digital-marketing-campaign-analysis'
FILE_PATH = 'data/clean_marketing_data.csv'
```

```python
# Membaca file CSV
with open('clean_marketing_data.csv', 'rb') as file:
    content = base64.b64encode(file.read()).decode('utf-8')
```

```python
# GitHub API
url = f'https://api.github.com/repos/{OWNER}/{REPO}/contents/{FILE_PATH}'

headers = {
    'Authorization': f'Bearer {GITHUB_TOKEN}',
    'Accept': 'application/vnd.github+json'
}
```

```python
# Memeriksa apakah file sudah tersedia
check = requests.get(url, headers=headers)

payload = {
    'message': 'Add Cleaned Marketing Dataset',
    'content': content
}

if check.status_code == 200:
    payload['sha'] = check.json()['sha']
elif check.status_code != 404:
    check.raise_for_status()
```

```python
# Upload ke GitHub
response = requests.put(url, headers=headers, json=payload)
response.raise_for_status()

print('Dataset berhasil diunggah ke GitHub!')
```
